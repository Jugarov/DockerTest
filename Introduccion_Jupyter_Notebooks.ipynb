# Introducción a Jupyter Notebooks

Un **Jupyter Notebook** es un documento interactivo que combina:

- texto y documentación en **Markdown**;
- código ejecutable;
- resultados de las ejecuciones;
- gráficos, tablas y otros elementos enriquecidos.

Es muy utilizado en **ciencia de datos, inteligencia artificial, programación y docencia**.

## 1. ¿Qué es una celda?

Un notebook está compuesto por **celdas**.

Las dos clases que utilizaremos principalmente son:

- **Markdown** → texto, títulos, listas, fórmulas y documentación.
- **Code** → código que será ejecutado por el kernel.

Una celda se puede ejecutar y su resultado queda asociado al notebook.

## 2. Markdown básico

Markdown permite escribir documentación de manera sencilla.

### Títulos

```markdown
# Título 1
## Título 2
### Título 3
```

### Énfasis

- **Negrita:** `**texto**`
- *Cursiva:* `*texto*`
- `Código`: `` `código` ``

### Listas

```markdown
- Elemento 1
- Elemento 2
  - Subelemento

1. Primer paso
2. Segundo paso
3. Tercer paso
```

### Enlaces

```markdown
[Texto del enlace](https://jupyter.org)
```

### Citas

```markdown
> Esta es una cita o una observación importante.
```

## 3. Fórmulas matemáticas

Markdown también permite escribir fórmulas utilizando **LaTeX**.

Una fórmula en línea:

$E = mc^2$

Una fórmula centrada:

$$
f(x) = x^2 + 2x + 1
$$

Esto resulta especialmente útil en notebooks científicos y de ingeniería.

## 4. Tablas

También podemos representar información mediante tablas:

| Elemento | Tipo | Ejemplo |
|---|---|---|
| Markdown | Texto | Títulos y explicaciones |
| Code | Código | Python |
| Kernel | Ejecución | Python 3 |
| Output | Resultado | Texto, números o gráficos |

Las tablas son útiles para resumir información de forma compacta.

## 5. Selección del kernel

El **kernel** es el proceso que ejecuta el código del notebook.

En un notebook de Python, normalmente seleccionaremos un kernel asociado a una instalación o entorno de Python.

En VS Code:

1. Abre un archivo `.ipynb`.
2. Busca el selector de **Kernel** en la parte superior derecha.
3. Haz clic sobre él.
4. Selecciona el entorno de Python que quieras utilizar.

Por ejemplo, puedes tener:

```text
Python 3.12
Python 3.11 (venv)
Python 3.10 (conda)
```

El kernel elegido determina qué versión de Python, paquetes y entorno estarán disponibles para el notebook.

> **Importante:** cambiar el kernel puede cambiar qué librerías están instaladas y cómo se ejecuta el código.

## 6. Código

Las celdas de código contienen instrucciones que serán ejecutadas por el **kernel**.

Por ejemplo, podemos realizar una operación sencilla:

````python
a = 10
b = 5

resultado = a + b
resultado
````

El valor que aparece debajo de la celda es el **resultado de su ejecución**.

También podemos utilizar `print()` para mostrar información:

````python
nombre = "Jupyter"
print("Hola,", nombre)
````

## 7. Variables y estado del notebook

El kernel mantiene el estado de las variables entre ejecuciones.

Ejecuta primero esta celda:

````python
mensaje = "Esta variable quedó almacenada en el kernel"
numero = 42
````

Ahora ejecuta la siguiente celda. Aunque no volvemos a definir las variables, siguen disponibles:

````python
print(mensaje)
print("El número es:", numero)
````

### ¿Qué ocurrió?

El notebook no ejecuta necesariamente todo el documento de arriba hacia abajo.

El **kernel mantiene un estado de ejecución**. Por eso es posible ejecutar una celda varias veces o ejecutar celdas en un orden diferente.

Esto es muy útil para experimentar, pero también puede generar confusión si el estado del kernel no coincide con el orden visual del notebook.

````python
producto = "Notebook"
precio = 100
cantidad = 3

total = precio * cantidad

print(f"Producto: {producto}")
print(f"Cantidad: {cantidad}")
print(f"Total: ${total}")
````

## 8. Reiniciar el kernel

Como el kernel mantiene variables en memoria, en ocasiones necesitamos reiniciarlo.

En VS Code podemos utilizar la opción **Restart Kernel**.

Al reiniciarlo:

- se eliminan las variables almacenadas;
- se reinicia el estado de ejecución;
- las celdas ya ejecutadas no conservan sus variables en memoria.

Después de reiniciar, si ejecutamos:

```python
print(mensaje)
```

obtendremos un error porque `mensaje` ya no existe.

> **Idea clave:** el contenido del notebook y el estado del kernel son conceptos diferentes.
